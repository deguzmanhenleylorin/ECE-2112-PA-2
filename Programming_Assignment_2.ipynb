**A. REPRODUCIBLE NORMALIZATION PROBLEM** - Create a reproducible random 5×5 integer ndarray named X.

```python
#Generating a 5x5 integer ndarray X
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
#Calculating the mean
mean = np.mean(X)
mean
```

```python
#Calculating the standard deviation
std = np.std(X)
std
```

```python
#Normalizing the array using the given formula in the file
X_normalized = (X - mean) / std
X_normalized
```

```python
#Normalized mean and standard deviation
print("\nMean of X_normalized:", np.mean(X_normalized))
print("Standard Deviation of X_normalized:", np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
```

**B. CUBES DIVISIBLE BY 4 PROBLEM** -  Create the first 100 positive integers, cube every element and reshape the result into a 10 × 10 ndarray named C.

---



```python
#The first 100 positive integers, cube every element and reshape the result into a 10 × 10 ndarray named C.
C = (np.arange(1, 101) ** 3).reshape(10, 10)
```

```python
#Using Boolean condition on C to select every cube value divisible by 4
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
#Displaying the shape of C and the number of selected elements
print("Shape of C:", C.shape)
print("\nSelected elements:", div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

**C. ABOVE-MEAN SQUARES PROBELM** - Create a 6 × 6 ndarray named S containing the squares of the first 36 positive integers in increasing row-major order.

```python
#A 6 × 6 ndarray named S containing the squares of the first 36 positive integers in increasing row-major order.
S = (np.arange(1, 37) ** 2).reshape(6, 6)
```

```python
#Calculating the mean
mean = np.mean(S)
```

```python
#Calculating above_mean
above_mean = S[S > mean]
```

```python
#Displaying S, mean, above_mean, and the number of selected elements.
print("Array S:\n", S)
print("\nMean of S (mean):", mean)
print("\nArray above_mean:", above_mean)
print("\nNumber of selected elements:", above_mean.size)
```

```python
np.save("above_mean.npy", above_mean)
```
